## 3. Frozen protocol

**Data:** BCI Competition IV 2a/BNCI2014_001. Source session runs 1–5 train, run 6 validates;
the complete target session is the E1/E2 test. E3 reserves target run 1 for calibration and
tests on runs 2–6 at *every* budget, including zero. No EOG classifier input; all artifact-marked
trials are retained and separately audited. Invalid/incomplete data raises an error.

**Timing/preprocessing:** the MATLAB `trial` marker is trial onset, **two seconds before the cue**.
The loader takes samples `[marker−1+500 : marker−1+1500]` at 250 Hz. Fourth-order Butterworth
4–38 Hz SOS filtering runs forward/backward separately within each recording run (odd padding,
SciPy default pad length). Polyphase anti-aliasing downsampling uses Kaiser β=5 to produce
22 × 500 input values. Channel standardization uses source-training epochs only. This is
offline preprocessing; it is not an online/causal BCI claim.

**Common spiking-family model:** temporal convolution (8 filters, 31 samples), spatial mixing
over 22 electrodes to 16 channels, average pooling by 4, GroupNorm(1), arctan/LIF dynamics;
then a 5-sample convolution to 32 channels, GroupNorm(1), and another LIF block. LIF leak=0.9,
threshold=1, zero initial state, subtractive previous-activity reset. LIF steps are 32 ms.
The final head uses mean activity over 125 steps. GroupNorm uses each trial's channel/time
statistics without running averages. No dropout or batch running statistics in this family.
Direct-hard uses binary forward spikes and the arctan derivative backward, including reset paths.

**EEGNet reference:** independent PyTorch implementation of EEGNet-8,2: 32-sample MI temporal
kernel, depthwise spatial filters, separable temporal convolution, BN, ELU, pooling 4/8,
dropout=0.5, spatial max-norm 1 and head max-norm 0.25. Adapted to 125 Hz and 500 samples;
PyTorch initialization and the common optimizer are used. This is an external reference,
not an architecture-matched conversion control.

**Optimization:** AdamW, batch 32 sampled with replacement, 100 epochs (200 for matched ordinary),
cosine learning rate to 1% of its initial value, no gradient clipping, AMP, or TF32. SAM uses
the same minibatch for both gradients; SAST uses independent ascent/descent sampling streams.
Hidden state resets on each pass. Both restore original parameters before the optimizer update.
Main seeds 0–4 share initial weights across the five matching architectures.

**Equal search budget:** four candidates per method/participant on pilot seed 1000, with the
same full per-method training budget. Every method searches LR {0.0003, 0.001}; SAM/SAST
also search ρ {0.05, 0.2} at weight decay 0.0001, while other methods search weight decay
{0.0001, 0.001}. Select using source-validation accuracy, then loss, then candidate order.
Select best native-forward checkpoint at 20 evenly spaced validation points; no early stopping.
All tuning and main training complete before target sessions are opened for inference.

**Calibration/probes:** global threshold grid {0.25, 0.5, 0.75, 1, 1.25, 1.5, 2, 3}; equal-score
ties favor the raw/source threshold, then smaller multiplier. Ridge α grid
{0.001, 0.01, 0.1, 1, 10, 100, 1000}; ties favor stronger regularization. Each mode/layer gets
a source-fitted feature scaler and ridge classifier. E3 readout fits source+calibration trials
with equal trial weights, fixed source scaling and α. Threshold-only and readout-only are
separate interventions with their own zero-label baselines.

**Statistics:** average seeds within each participant first. Report 10,000 paired participant
bootstrap resamples (seed 20261012); nine participants are the inferential units. Primary:
calibrated SAST minus calibrated matched ordinary balanced accuracy. CKA alone is not evidence
of preserved decoding. A null result is not an equivalence result. Undefined CKA stays missing.
